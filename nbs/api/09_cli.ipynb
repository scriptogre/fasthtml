# Command Line Tools

```python
#| export
from fastcore.utils import *
from fastcore.script import call_parse, bool_arg
from argparse import ArgumentParser, BooleanOptionalAction
from subprocess import check_output, run

import json
```

```python
#| export
@call_parse
def railway_link():
    "Link the current directory to the current project's Railway service"
    j = json.loads(check_output("railway status --json".split()))
    prj = j['id']
    idxpath = 'edges', 0, 'node', 'id'
    env = nested_idx(j, 'environments', *idxpath)
    svc = nested_idx(j, 'services', *idxpath)

    cmd = f"railway link -e {env} -p {prj} -s {svc}"
    res = check_output(cmd.split())
```

```python
#| export
def _run(a, **kw):
    print('#', ' '.join(a))
    run(a)
```

```python
#| export
@call_parse
def railway_deploy(
    name:str, # The project name to deploy
    mount:bool_arg=True # Create a mounted volume at /app/data?
):
    """Deploy a FastHTML app to Railway"""
    nm,ver = check_output("railway --version".split()).decode().split()
    assert nm.startswith('railway'), f'Unexpected railway version string: {nm}'
    if ver2tuple(ver)<(3,8): return print("Please update your railway CLI version to 3.8 or higher")
    cp = run("railway status --json".split(), capture_output=True)
    if not cp.returncode:
        print("Checking deployed projects...")
        project_name = json.loads(cp.stdout.decode()).get('name')
        if project_name == name: return print("This project is already deployed. Run `railway open`.")
    reqs = Path('requirements.txt')
    if not reqs.exists(): reqs.write_text('python-fasthtml')
    _run(f"railway init -n {name}".split())
    _run(f"railway up -c".split())
    _run(f"railway domain".split())
    railway_link.__wrapped__()
    if mount: _run(f"railway volume add -m /app/data".split())
    _run(f"railway up -c".split())
```

## Commands

The `fasthtml` and `fh` commands group related operations without removing the existing standalone commands.

```python
#| export
def main(argv=None, _link=None, _deploy=None):
    "Run the FastHTML command line interface."
    parser = ArgumentParser()
    commands = parser.add_subparsers(required=True)
    railway = commands.add_parser('railway', help='Manage Railway deployments')
    actions = railway.add_subparsers(dest='action', required=True)
    actions.add_parser('link', help='Link the current Railway project')
    deploy = actions.add_parser('deploy', help='Deploy a FastHTML app to Railway')
    deploy.add_argument('name', help='Project name')
    deploy.add_argument('--mount', action=BooleanOptionalAction, default=True,
                        help='Create a mounted volume at /app/data')
    args = parser.parse_args(argv)
    if args.action == 'link': return (_link or railway_link.__wrapped__)()
    return (_deploy or railway_deploy.__wrapped__)(args.name, args.mount)
```

```python
from fastcore.test import test_eq
```

```python
calls = []
main(['railway', 'link'], _link=lambda: calls.append(('link',)))
main(['railway', 'deploy', 'demo', '--no-mount'],
     _deploy=lambda name, mount: calls.append(('deploy', name, mount)))
test_eq(calls, [('link',), ('deploy', 'demo', False)])
calls
```
